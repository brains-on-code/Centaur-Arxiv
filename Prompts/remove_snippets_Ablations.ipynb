```python
import pandas as pd
import os
groups =  ["Katzmarski12"]

for group in groups: 
    print(group)
    snippets = []
    for file in os.listdir(f"{group}\\snippets"):
        with open (f"{group}\\snippets\\{file}", "r") as f:
            snippets.append(f.read().replace("\n", "\r\n"))
    for file in os.listdir(f"{group}\\csv_files\\whole_runs"):
        _,_,_,num = file.split("_")
        old_df = pd.read_csv(f"{group}\\csv_files\\whole_runs\\{file}")
        new_text = []
        new_ind = [i for i in range(1,len(old_df)+1)]
        new_exp = [group] * len(old_df)
        for i in range(len(old_df)):
            text = old_df["text"].iloc[i]
            match(group): 
                case "Medeiros19":
                    for snip in snippets: 
                        text = text.replace(snip, "You see a pair of code snippets.")
                case "Ajami18": 
                    for snip in snippets: 
                        text = text.replace(snip, "You see a code snippet.")
                case "Börstler16": 
                    for snip in snippets: 
                        text = text.replace(snip, "\nYou see a code snippet.\n")
                    cts = []
                    for file in os.listdir(f"{group}\\cloze_tests"):
                        with open (f"{group}\\cloze_tests\\{file}", "r") as f:
                            cts.append(f.read().replace("\n", "\r\n"))
                    for ct in cts: 
                        text = text.replace(ct, "\nYou see a cloze test of the prior snippet.\n")
                    
                case _ :
                    for snip in snippets: 
                        text = text.replace(snip, "You see a code snippet.\n")
            new_text.append(text)
        new_df = pd.DataFrame({"text": new_text, "participant": new_ind, "experiment": new_exp})
        os.makedirs(f"{group}\\csv_files\\ablations\\no_snippets\\", exist_ok=True)
        new_df.to_csv(f"{group}\\csv_files\\ablations\\no_snippets\\{group}_no_snippet_{num}", index = False)
    for file in os.listdir(f"{group}\\csv_files\\ablations\\no_order"):
        new_name = file.replace("no_order", "no_order_no_snippet")
        old_df = pd.read_csv(f"{group}\\csv_files\\ablations\\no_order\\{file}")
        new_text = []
        new_ind = [i for i in range(1,len(old_df)+1)]
        new_exp = [group] * len(old_df)
        for i in range(len(old_df)):
            text = old_df["text"].iloc[i]
            match(group): 
                case "Medeiros19":
                    for snip in snippets: 
                        text = text.replace(snip, "You see a pair of code snippets.\n")
                case "Ajami18": 
                    for snip in snippets: 
                        text = text.replace(snip, "You see a code snippet.")
                case "Börstler16": 
                    for snip in snippets: 
                        text = text.replace(snip, "\nYou see a code snippet.\n")
                    cts = []
                    for file in os.listdir(f"{group}\\cloze_tests"):
                        with open (f"{group}\\cloze_tests\\{file}", "r") as f:
                            cts.append(f.read().replace("\n", "\r\n"))
                    for ct in cts: 
                        text = text.replace(ct, "\nYou see a cloze test of the prior snippet.\n")
                case _ :
                    for snip in snippets: 
                        text = text.replace(snip, "You see a code snippet.\n")
            new_text.append(text)
        new_df = pd.DataFrame({"text": new_text, "participant": new_ind, "experiment": new_exp})
        os.makedirs(f"{group}\\csv_files\\ablations\\no_order_no_snippets\\", exist_ok=True)
        new_df.to_csv(f"{group}\\csv_files\\ablations\\no_order_no_snippets\\{new_name}", index = False)
```
