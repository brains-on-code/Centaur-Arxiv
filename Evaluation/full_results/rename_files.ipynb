```python
import os
```

```python
path = "Llama/Buse10_no_order_no_snippets_no_desc/"

for file in os.listdir(path):
    parts = file.split("_")
    if parts[-3] == "snippet": 
        new_name = ""
        id = len(parts)-3
        for i,part in enumerate(parts): 
            if i == id: continue
            else: 
                new_name += part
                if i != len(parts)-1:
                    new_name += "_"
        os.rename(f"{path}/{file}", f"{path}/{new_name}")
```
