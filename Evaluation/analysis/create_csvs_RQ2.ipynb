```python
import pandas as pd
```

```python
studies = ["Ajami18", "Bauer19", "Börstler16", "Buse10", "Gopstein17a", "Gopstein17b", "Medeiros19", "Siegmund17", "Peitek18", "overall"]
def create_csv(category):
    results_dict = {"study": [], "RQ1": [], "RQ2_no_order": [], "RQ2_no_snippets": [], "RQ2_no_order_no_snippets": [], "RQ2_no_order_no_desc": [], "RQ2_no_snippets_no_desc": [], "RQ2_no_desc": [], "RQ2_no_order_no_snippets_no_desc":[]}
    for study in studies:
        results_dict["study"].append(study) 
        for type in ["RQ1", "RQ2_no_order", "RQ2_no_snippets", "RQ2_no_order_no_snippets", "RQ2_no_desc", "RQ2_no_order_no_desc", "RQ2_no_order_no_snippets_no_desc", "RQ2_no_snippets_no_desc"]:
            results_df = pd.read_csv(f"results/results_{type}.csv")
            study_df = results_df[results_df["study"]==study]
            res = study_df[category].iloc[0]
            results_dict[type].append(res)
    print(category)
    print(results_dict)
    df_res = pd.DataFrame(results_dict)
    df_res.to_csv(f"results/RQ2/{category}.csv", index=False)
for category in ["cohen_d", "pvalue", "Centaur_mean", "Centaur_std", "Llama_mean", "Llama_std"]:
    create_csv(category)
```
