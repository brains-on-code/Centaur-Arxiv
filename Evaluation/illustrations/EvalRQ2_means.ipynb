```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
effect_sizes = pd.read_csv("results/RQ2/Centaur_mean.csv")
result_df = {"study": [], "Sequence;Code;Desc": [], "Sequence;Code": [], "Sequence;Desc": [], "Code;Desc": [], "Code": [], "Sequence": [], "Desc": [], "Base": []}
for i in range(len(effect_sizes)): 
    result_df["study"].append(effect_sizes["study"].iloc[i])
    result_df["Sequence;Code;Desc"].append(effect_sizes["RQ1"].iloc[i])
    result_df["Sequence;Code"].append(effect_sizes["RQ2_no_desc"].iloc[i])
    result_df["Sequence;Desc"].append(effect_sizes["RQ2_no_snippets"].iloc[i])
    result_df["Code;Desc"].append(effect_sizes["RQ2_no_order"].iloc[i])
    result_df["Code"].append(effect_sizes["RQ2_no_order_no_desc"].iloc[i])
    result_df["Sequence"].append(effect_sizes["RQ2_no_snippets_no_desc"].iloc[i])
    result_df["Desc"].append(effect_sizes["RQ2_no_order_no_snippets"].iloc[i])
    result_df["Base"].append(effect_sizes["RQ2_no_order_no_snippets_no_desc"].iloc[i])
res_df = pd.DataFrame(result_df)
res_df.to_csv("results/RQ2/substractive_effects.csv", index = False)
```

```python
centaur_heatmap_df = res_df.set_index("study")
plt.figure(layout = "tight")
sns.heatmap(centaur_heatmap_df, annot = True, cmap="Reds", vmax = 10, vmin= 0, fmt= ".2f")
plt.title("Mean neg. Log Likelihood for Centaur across Ablations")
plt.xticks(plt.xticks()[0], ["S+C+D", "S+C", "S+D", "C+D", "C", "S", "D", "NoInfo"], rotation = 0)
plt.yticks(plt.yticks()[0], [f"S{i}" for i in range(1,10)] + ["Overall"])
plt.xlabel("Ablation")
plt.savefig("images/heatmap_centaur.pdf")
```
